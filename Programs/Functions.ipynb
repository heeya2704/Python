```python
# create a function square which print square of given number
def square(num):
    return num ** 2
number = int(input("Enter a number: "))
result = square(number)
print(f"The square of {number} is {result}.")
```

```python
def odd_even(num):
    if num % 2 == 0 :
        return "Even"
    else:        
        return "Odd"
number = int(input("Enter a number: "))
result = odd_even(number)
print(f"The number {number} is {result}.")
```

```python
# create a function square which print square of given number and call the function in list
def square(num):
    return num ** 2
number = int(input("Enter a number: "))
result = square(number)
lst=[result]
print(f"The square of {number} is {result} and {lst}.")
```

```python
# create a function square which print square of given number and call the function from list and also store result in list
def square(num):
    return num ** 2
lst = [3, 4, 5]
for i in lst:
    result = square(i)
    print(f"The square of {i} is {result}.")
```

```python
# create afunction for power and exponent take input from the user and print the result
def power(a, b):
    return a ** b
base = int(input("Enter the base: "))
pwr = int(input("Enter the power: "))
result = power(base, pwr)
print(f"{base} raised to the power of {pwr} is {result}.")
```

```python
def checkpositive(num):
    if num>0:
        return True
    else:
        return False
def square(num):
    return num ** 2
number = int(input("Enter a number: "))
while True:
    if checkpositive(number):
        result = square(number)
        print(f"The square of {number} is {result}.")
        break
    else:
        print("Please enter a positive number.")
        number = int(input("Enter a number: "))
```

```python
# create a function to add to two list into third list and print the result
def merge_lists(list1, list2):
    return list1 + list2
list1 = [1, 2, 3]
list2 = [4, 5, 6]
list3 = []
list3 = merge_lists(list1, list2)
print(f"The merged list is: {list3}.")
```

```python
# create a function to add two dictionary in third dictionary and print the result
def merge_dicts(dic1, dic2):
    return {**dic1, **dic2}
dic1 = {1: 'a', 2: 'b', 3: 'c'}
dic2 = {4: 'd', 5: 'e', 6: 'f'}
dic3 = {}
dic3 = merge_dicts(dic1, dic2)
print(f"The merged dictionary is: {dic3}.")
```
