```python
# count total number of even and odd digits in the number and also find sum of digits
number = int (input("Enter Number : "))
even_count = 0
odd_count = 0
sum = 0
num1 = number
while num1!=0 :
    rem = num1 % 10
    sum+=rem
    if rem%2 == 0:
        even_count+=1
    else:
        odd_count+=1
    num1 = num1//10
print(f'{number} has {even_count} even digits and {odd_count} odd digits and sum of the digits is {sum}')
```

### Menu driven programs
1. factorial
2. count of digits
3. sum of digits
4. prime number
5. exit

```python
# Menu driven programs
print("1. find factorial number")
print("2. count number of digits")
print("3. sum of the digits present in the number") 
print("4. find prime number")
print("5. exit")
while True:
    choice = int(input("Enter your choice : "))
    match choice:
        case 1:
            number = int(input("Enter Number : "))
            factorial = 1
            for i in range(1, number + 1):
                factorial = factorial * i
            print("Factorial of", number, "is", factorial)
        case 2:
            number = int (input("Enter Number : "))
            count = 0 
            num1 = number
            while num1!=0 :
                rem = num1 % 10
                print (rem)
                count+=1
                num1 = num1//10
            print(f'{number} has {count} digits')
        case 3:
            number = int (input("Enter Number : "))
            sum = 0
            num1 = number
            while num1!=0 :
                rem = num1 % 10
                sum+=rem
                num1 = num1//10
            print(f'sum of the digits present in {number} is {sum}')
        case 4:
            number = int (input("Enter Number : "))
            if number > 1:
                for i in range(2, number):
                    if (number % i) == 0:
                        print(number, "is not a prime number")
                        break
                    else:
                        print(number, "is a prime number")
            else:
                print(number, "is not a prime number")
        case 5:
            print("Exiting the program...")
            break
        case _:
            print("Invalid choice")
```

### Armstrong number

```python
# armstrong number
number = int (input("Enter Number : "))
sum = 0
num1 = number
while num1!=0 :
    rem = num1 % 10
    sum+=rem**3
    num1 = num1//10
if sum == number:
    print(number, "is an armstrong number")
else:
    print(number, "is not an armstrong number")
```

```python
# nested loop - each row have one loop and each column have another loop
num = int (input("Enter Number : "))
for i in range(num*2-1):
    if i<num:
        for j in range(i,num):
            print("*", end=" ")
    else:
        for j in range((i-num)+2):
            print("*", end=" ")
    print()
```
