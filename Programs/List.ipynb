### In built methods in list

1. Append() - add the elements into the list at last

```python
list_number = [23, 45, 67, 89, 12]
print(f"Before append method: {list_number}")
list_number.append(900)
print(f"After append method: {list_number}")
list1 = [90,34,56]
list_number.append(list1)
print(f"After append method with list: {list_number}")
```

2. Extend() - add the individual element in the list from the another list

```python
list_number = [23, 45, 67, 89, 12]
print(f"Before append method: {list_number}")
list_number.append(900)
print(f"After append method: {list_number}")
list1 = [90,34,56]
list_number.extend(list1)
print(f"After extend method with list: {list_number}")
```

```python
# append vs extend
# extend the city names in the list like kolkata then 
# output will be like this ['Mumbai', 'Delhi', 'Bangalore', 'Pune', 'Chennai', 'Kolkata']
list_city = ["Mumbai", "Delhi", "Bangalore", "Pune"]
list_city.append("Chennai")
list1 = ["Kolkata"]
list_city.extend(list1)
print(list_city)
```

```python
# shopping list
# menu driven program
# 1. add item to the shopping list - which item : grocery
# 2. update item to the shopping list - press 1 for item name and press 2 for index number
#       if user press 1 then ask for the item name and add that item to the shopping list 
#       eg: washing powder then output will be your list has been updated
# 3. delete item from the shopping list
# 4. search item in the shopping list
# 5. display the shopping list
# 6. exit
shopping_list = []
while True:
    print("1. Add item to the shopping list")
    print("2. Update item in the shopping list")
    print("3. Delete item from the shopping list")
    print("4. Search item in the shopping list")
    print("5. Display the shopping list")
    print("6. Exit")
    choice = input("Enter your choice: ")
    
    if choice == '1':
        item = input("Enter the item name to add: ")
        shopping_list.append(item)
        print(f"{item} has been added to the shopping list.")
    
    elif choice == '2':
        update_choice = input("Press 1 to update by item name or Press 2 to update by index number: ")
        if update_choice == '1':
            old_item = input("Enter the item name to update: ")
            if old_item in shopping_list:
                new_item = input("Enter the new item name: ")
                index = shopping_list.index(old_item)
                shopping_list[index] = new_item
                print(f"{old_item} has been updated to {new_item}.")
            else:
                print(f"{old_item} is not in the shopping list.")
        
        elif update_choice == '2':
            index = int(input("Enter the index number to update: "))
            if 0 <= index < len(shopping_list):
                new_item = input("Enter the new item name: ")
                old_item = shopping_list[index]
                shopping_list[index] = new_item
                print(f"{old_item} has been updated to {new_item}.")
            else:
                print("Invalid index number.")
    
    elif choice == '3':
        item_to_delete = input("Enter the item name to delete: ")
        if item_to_delete in shopping_list:
            shopping_list.remove(item_to_delete)
            print(f"{item_to_delete} has been deleted from the shopping list.")
        else:
            print(f"{item_to_delete} is not in the shopping list.")
    
    elif choice == '4':
        item_to_search = input("Enter the item name to search: ")
        if item_to_search in shopping_list:
            print(f"{item_to_search} is present in the shopping list.")
        else:
            print(f"{item_to_search} is not present in the shopping list.")
    
    elif choice == '5':
        print("Shopping List:")
        for item in shopping_list:
            print(f"- {item}")
    elif choice == '6':
        print("Exiting the program. Goodbye!")
        break
    else:
        print("Invalid choice. Please try again.")
```

```python
# menu driven program for employee management system where we can add employee details like id, name, age,email, salary, department 
# and id should be from 101 to 200 and name should be in string and age should be in between 18 to 60 and email should be in correct format 
# and salary should be in between 30000 to 100000 and department should be in HR, IT, Finance, Marketing 
# we can also update the employee details with id nd show ids and delete the employee details with the id and search the employee details and 
# display the employee details and exit the program
# search employee details from salary it show all the employee details whose salary is greater than or equal to the salary entered by the user
# if not no record found
emp_data ={
    101: {"name": "Alice", "age": 30, "email": "alice@example.com", "salary": 50000, "department": "HR"},
    102: {"name": "Bob", "age": 35, "email": "bob@example.com", "salary": 60000, "department": "IT"},
    103: {"name": "Charlie", "age": 40, "email": "charlie@example.com", "salary": 70000, "department": "Finance"},
    104: {"name": "David", "age": 28, "email": "david@example.com", "salary": 55000, "department": "Marketing"}
}
while True:
    print("1. Add employee details")
    print("2. Update employee details")
    print("3. Delete employee details")
    print("4. Search employee details")
    print("5. Search employee details by salary")
    print("6. Search employee details by department")
    print("7. Display employee details")
    print("8. Exit")
    choice = input("Enter your choice: ")
    
    if choice == '1':
        emp_id = int(input("Enter employee ID: "))
        name = input("Enter employee name: ")
        age = int(input("Enter employee age: "))
        email = input("Enter employee email: ")
        salary = int(input("Enter employee salary: "))
        department = input("Enter employee department: ")
        emp_data[emp_id] = {"name": name, "age": age, "email": email, "salary": salary, "department": department}
        print(f"Employee {name} has been added.")
    
    elif choice == '2':
        emp_id = int(input("Enter employee ID to update: "))
        if emp_id in emp_data:
            name = input("Enter new name (leave blank to keep current): ")
            age_input = input("Enter new age (leave blank to keep current): ")
            email = input("Enter new email (leave blank to keep current): ")
            salary_input = input("Enter new salary (leave blank to keep current): ")
            department = input("Enter new department (leave blank to keep current): ")
            
            if name:
                emp_data[emp_id]["name"] = name
            if age_input:
                emp_data[emp_id]["age"] = int(age_input)
            if email:
                emp_data[emp_id]["email"] = email
            if salary_input:
                emp_data[emp_id]["salary"] = int(salary_input)
            if department:
                emp_data[emp_id]["department"] = department
            
            print(f"Employee ID {emp_id} has been updated.")
        else:
            print(f"Employee ID {emp_id} not found.")
    
    elif choice == '3':
        emp_id = int(input("Enter employee ID to delete: "))
        if emp_id in emp_data:
            del emp_data[emp_id]
            print(f"Employee ID {emp_id} has been deleted.")
        else:
            print(f"Employee ID {emp_id} not found.")
    
    elif choice == '4':
        emp_id = int(input("Enter employee ID to search: "))
```
