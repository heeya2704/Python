Arithmetic Operators

% - modulus (give us the reminder)

// - floor division 

** - exponential (power - raise to function)

```python
# Accept 2 numbers from user 1st as a base value and 2nd as an exponent value and print the result of base value raised to the power of exponent value.

base = int(input("Enter the base value: "))
exponent = int(input("Enter the exponent value: "))
result = base ** exponent
print(f"The result of {base} raised to the power of {exponent} is: {result}")
```

```python
# same with the floor division and modulus operator

num1 = int(input("Enter the first number: "))
num2 = int(input("Enter the second number: "))
floor_division_result = num1 // num2
modulus_result = num1 % num2
print(f"The result of floor division of {num1} by {num2} is: {floor_division_result}")
print(f"The result of modulus of {num1} by {num2} is: {modulus_result}")
```

### Relational Operator

 -- > , <, >=, <= , !=, ==

returns output True/false

```python
num1 = 130
num2 = 100
print(num1 > num2)  # True
print(num1 < num2)  # False
print(num1 >= num2)  # True
print(num1 <= num2)  # False
print(num1 == num2)  # False
print(num1 != num2)  # True
```

Logical operators

AND , OR , NOT

```python
num1 = 100
num2 =17
print(num1 > num2 and num1>100)  # False
print(num1 > num2 or num1>100)  # True
print(not(num1 > num2))  # False
```

```python
# enter a mark and check the grade of the student
mark = int(input("Enter the mark: "))
if mark >= 85 and mark <= 100:
    print("Grade: A")
elif mark >= 70 and mark < 85:
    print("Grade: B")
elif mark >= 55 and mark < 70:
    print("Grade: C")
elif mark >= 35 and mark < 55:
    print("Grade: D")
else:
    print("Grade: F")
```

```python
# enter a age from the user and check if the person is a 0-2 infant, 3-12 child, 13-18 teenager, 
# 19-50 adult, 51-70 senior citizen and above 70 super senior citizen. but user can enter only in the range of 0-105. 
# if user enters age less than 0 or greater than 105 then print invalid age. use if within if statement called as a nested if.
age = int(input("Enter your age: "))
if age >= 0 or age <= 105:
    if age >= 0 and age <= 2:
        print("You are an infant.")
    elif age >= 3 and age <= 12:
        print("You are a child.")
    elif age >= 13 and age <= 18:
        print("You are a teenager.")
    elif age >= 19 and age <= 50:
        print("You are an adult.")
    elif age >= 51 and age <= 70:
        print("You are a senior citizen.")
    else:
        print("You are a super senior citizen.")
else:
    print("Invalid age.")
```

```python
# check the user age and weight for a person to be eligible for blood donation. 
# the age should be between 18 and 65 and weight should be greater than 55 kg. use nested if statement.
age = int(input("Enter your age: "))
if age >= 18 and age <= 65:
    weight = float(input("Enter your weight in kg: "))
    if weight > 55:
        print("You are eligible for blood donation.")
    else:
        print("You are not eligible for blood donation due to weight.")
else:
    print("You are not eligible for blood donation due to age doesnot lie in the age criteria.")
```

### Assignment Operators 

+= , -= , *= , /=

```python
# increment and decrement of the a variable using the assignment operator
a=10
a += 1  # increment by 1
print(f"Updated value of a after incrementing by 1: {a}")
a -= 1  # decrement by 1
print(f"Updated value of a after decrementing by 1: {a}")
a *= 2  # multiply by 2
print(f"Updated value of a after multiplying by 2: {a}")
a /= 2  # divide by 2
print(f"Updated value of a after dividing by 2: {a}")
```
