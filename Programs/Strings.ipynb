## Built-in methods

1. upper()

```python
name= input("Enter your name: ")
print(f"{name} in upper case : {name.upper()}")
```

2. lower()


```python
name= input("Enter your name: ")
print(f"{name} in lower case : {name.lower()}")
```

3. title()

```python
name= input("Enter your name: ")
print(f"{name} in title : {name.title()}")
```

4. capitalise()

```python
name= input("Enter your name: ")
print(f"{name} in capitalise case : {name.capitalize()}")
```

5. endswith()

```python
# check email is valid or not using endwith() method
email = input("Enter your email : ")
if email.endswith("@gmail.com"):
    print(f"Valid Email: {email}")
else:
    print(f"Invalid Email: {email}")
```

6. startswith()

```python
# check number is valid or not using startswith() method
contact_number = input("Enter your number : ")
if contact_number.startswith("91"):
    print(f"Valid Number: {contact_number}")
else:
    print(f"Invalid Number: {contact_number}")
```

7. len()

```python
# check number is valid or not using startswith() method and also with 6,7,8,9 and also with length 12
contact_number = input("Enter your number : ")
if contact_number.startswith("91"):
    if contact_number.startswith(("6", "7", "8", "9")) and len(contact_number) == 12:
        print(f"Valid Number: {contact_number}")
    else:
        print(f"Invalid Number: {contact_number}")
else:
    print(f"Invalid Number: {contact_number}")
```

8. str()

```python
# check number is valid or not using startswith() method with int input
contact_number = int(input("Enter your number : "))
contact_number = str(contact_number)
if contact_number.startswith("91"):
    print(f"Valid Number: {contact_number}")
else:
    print(f"Invalid Number: {contact_number}")
```

9. count()

```python
name = input("Enter your name: ")
i = name.count("a")
print(f"Number of a in {name} is : {i}")
```

```python
name = input("Enter your name: ")
i = name.count("is")
print(f"Number of 'is' word in '{name}' is : {i}")
```

```python
# count number of letters a after spliting the string with " " space
name = input ("Enter your name: ")
ans = name.split(" ") 
print(ans)
for i in ans:
    print(f"Number of a in '{i}' is : {i.count('a')}")
```

```python
name = input ("Enter your name: ")
ans = name.split(" ") 
print(f'{ans} contains {len(ans)} words and length of string is {len(name)}')
for i in ans:
    print(f"Number of a in '{i}' is : {i.count('a')}")
```
